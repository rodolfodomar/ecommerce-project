```python
# --- BOILERPLATE: INITIALIZE SPARK SESSION ---
import findspark
findspark.init()

from pyspark.sql import SparkSession
from pyspark.sql.types import StructType, StructField, StringType, TimestampType

# Initialize SparkSession
spark = SparkSession.builder \
    .appName("Olist Data Exploration") \
    .master("local[*]") \
    .getOrCreate()

# --- STEP 1: LOAD THE ORDERS DATA ---

# Define the schema to ensure data consistency and improve performance.
orders_schema = StructType([
    StructField("order_id", StringType(), True),
    StructField("customer_id", StringType(), True),
    StructField("order_status", StringType(), True),
    StructField("order_purchase_timestamp", TimestampType(), True),
    StructField("order_approved_at", TimestampType(), True),
    StructField("order_delivered_carrier_date", TimestampType(), True),
    StructField("order_delivered_customer_date", TimestampType(), True),
    StructField("order_estimated_delivery_date", TimestampType(), True)
])

# Define the path to your raw data inside the container.
# The project folder is mounted at /home/jovyan/workspace
data_path = "/home/jovyan/workspace/data/raw/"

# Read the CSV file using the defined schema
orders_df = spark.read.csv(
    f"{data_path}olist_orders_dataset.csv",
    header=True,
    schema=orders_schema
)

# --- STEP 2: EXPLORE THE DATA ---

# Show the first 5 rows of the DataFrame
print("Displaying first 5 rows of the orders table:")
orders_df.show(5)

# Print the schema to verify data types
print("Orders table schema:")
orders_df.printSchema()

# Get a summary of the DataFrame
print(f"Total number of orders: {orders_df.count()}")
```
